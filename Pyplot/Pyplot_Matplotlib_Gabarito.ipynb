```python
import numpy as np
import matplotlib.pyplot as plt
print("Bibliotecas carregadas!")
```

## 1 — Gráfico de linhas (`plot`)

### Exercício 1.1


```python
dias = ['Seg','Ter','Qua','Qui','Sex','Sáb','Dom']
temperaturas = [22,24,21,23,26,28,25]

plt.figure(figsize=(9,5))
plt.plot(dias, temperaturas, marker='o')
plt.title('Temperatura média da semana')
plt.xlabel('Dia da semana')
plt.ylabel('Temperatura (°C)')
plt.grid(True)
plt.show()
```

### Exercício 1.2


```python
periodos = range(1,7)
a = [100,120,140,130,170,200]
b = [180,170,160,150,140,130]

plt.figure(figsize=(9,5))
plt.plot(periodos, a, marker='o', linestyle='-', label='Produto A')
plt.plot(periodos, b, marker='s', linestyle='--', label='Produto B')
plt.title('Tendência dos produtos')
plt.xlabel('Período')
plt.ylabel('Valor')
plt.legend()
plt.grid(True)
plt.show()
```

**Resposta:** o Produto A apresenta melhor tendência, pois cresce de 100 para 200. O Produto B apresenta tendência decrescente, de 180 para 130.

## 2 — Barras (`bar` / `barh`)

### Exercício 2.1


```python
cidades = ['Cidade A','Cidade B','Cidade C','Cidade D','Cidade E']
pop = [120,350,210,580,420]

plt.figure(figsize=(8,5))
plt.bar(cidades, pop, color='teal', edgecolor='black')
plt.title('População estimada')
plt.xlabel('Cidade')
plt.ylabel('População (milhares)')
plt.show()
```

### Exercício 2.2


```python
departamentos = [
'Recursos Humanos e Talentos',
'Controladoria e Finanças Corporativas',
'Pesquisa, Desenvolvimento e Inovação',
'Tecnologia da Informação e Infraestrutura',
'Marketing Estratégico e Comunicação',
'Operações Industriais e Logística',
'Atendimento ao Cliente e Suporte',
'Jurídico e Governança Corporativa']
valores = [72,85,91,95,80,88,76,70]

fig, ax = plt.subplots(1,2,figsize=(18,7))
ax[0].bar(departamentos,valores)
ax[0].tick_params(axis='x',rotation=90)
ax[0].set_title('bar()')
ax[1].barh(departamentos,valores)
ax[1].set_title('barh()')
plt.tight_layout()
plt.show()
```

## 3 — Dispersão (`scatter`)

### Exercício 3.1


```python
horas = [1,2,3.5,4,5.5,6,7.5,8,9,10]
notas = [3.0,4.0,5.5,5.0,6.8,7.2,8.0,7.8,9.2,9.5]

plt.figure(figsize=(8,5))
plt.scatter(horas,notas,s=60,alpha=0.7)
plt.title('Horas de estudo × Nota')
plt.xlabel('Horas de estudo')
plt.ylabel('Nota')
plt.grid(True)
plt.show()
```

### Exercício 3.2


```python
np.random.seed(42)
x1 = np.random.uniform(0,10,100)
y1 = 2*x1 + np.random.normal(0,2,100)
x2 = np.random.uniform(0,10,100)
y2 = np.random.uniform(0,10,100)

fig,ax=plt.subplots(1,2,figsize=(13,5))
ax[0].scatter(x1,y1,alpha=.7); ax[0].set_title('Conjunto 1')
ax[1].scatter(x2,y2,alpha=.7); ax[1].set_title('Conjunto 2')
plt.tight_layout(); plt.show()
```

**Resposta:** o Conjunto 1 mostra correlação positiva. No Conjunto 2, X e Y são independentes; portanto, não se espera relação linear sistemática e a correlação tende a ficar próxima de zero.

## 4 — Histograma (`hist`)

### Exercício 4.1


```python
notas = [4.0,4.5,5.0,5.0,5.5,5.8,6.0,6.0,6.2,6.5,6.8,7.0,7.0,7.2,7.5,7.8,8.0,8.0,8.2,8.5,8.8,9.0,9.2,9.5,10.0]

fig,ax=plt.subplots(1,2,figsize=(12,5))
ax[0].hist(notas,bins=4,edgecolor='black'); ax[0].set_title('4 bins')
ax[1].hist(notas,bins=10,edgecolor='black'); ax[1].set_title('10 bins')
plt.tight_layout(); plt.show()
```

**Resposta:** com 4 bins vemos um resumo mais geral; com 10 bins aparecem mais detalhes e pequenas variações da distribuição.

### Exercício 4.2


```python
notas_outlier = notas + [20.0]
fig,ax=plt.subplots(1,2,figsize=(12,5))
ax[0].hist(notas_outlier,bins=10,edgecolor='black')
ax[0].set_title('Histograma')
ax[1].boxplot(notas_outlier)
ax[1].set_title('Boxplot')
plt.tight_layout(); plt.show()
```

**Resposta:** no histograma, 20 aparece em uma faixa distante e amplia a escala. No boxplot, o valor é destacado além dos bigodes, facilitando sua identificação como outlier.

## 6 — Boxplot

### Exercício 6.1


```python
dados=np.array([12,14,15,15,16,17,18,19,19,20,22,24,38])
q1=np.percentile(dados,25)
mediana=np.percentile(dados,50)
q3=np.percentile(dados,75)
iqr=q3-q1
limite=q3+1.5*iqr

print('Q1:',q1)
print('Mediana:',mediana)
print('Q3:',q3)
print('IQR:',iqr)
print('Limite superior:',limite)
print('Outlier(s):',dados[dados>limite])

plt.boxplot(dados,vert=False)
plt.title('Boxplot')
plt.show()
```

**Resposta:** pela regra `Q3 + 1.5 × IQR`, o valor **38** é outlier.

### Exercício 6.2


```python
np.random.seed(42)
alfa=np.random.normal(100,6,120)
beta=np.random.normal(120,2,120)
gama=np.concatenate([np.random.normal(95,5,105),
                     np.random.exponential(18,15)+110])

plt.figure(figsize=(9,5))
plt.boxplot([alfa,beta,gama],tick_labels=['Alfa','Beta','Gama'])
plt.ylabel('Tempo (ms)')
plt.title('Tempos de resposta')
plt.show()
```

**Resposta:** Beta é o mais previsível porque apresenta menor dispersão e IQR mais estreito. Gama apresenta cauda superior e múltiplos valores extremos.

## 8 — Violino (`violinplot`)

### Exercício 8.1


```python
operacional=[2200,2300,2400,2500,2550,2600,2700,2800]
tecnologia=[3500,3800,4200,4500,5000,5200,6000,7500]

plt.figure(figsize=(8,5))
plt.violinplot([operacional,tecnologia],showmeans=True,showmedians=True)
plt.xticks([1,2],['Operacional','Tecnologia'])
plt.ylabel('Salário (R$)')
plt.title('Salários de entrada')
plt.show()
```

### Exercício 8.2


```python
np.random.seed(42)
bimodal=np.concatenate([np.random.normal(30,5,300),
                        np.random.normal(70,6,300)])
fig,ax=plt.subplots(1,2,figsize=(11,5))
ax[0].boxplot(bimodal); ax[0].set_title('Boxplot')
ax[1].violinplot(bimodal,showmeans=True,showmedians=True); ax[1].set_title('Violino')
plt.tight_layout(); plt.show()
```

**Resposta:** o boxplot resume quartis, mediana e extremos. O violino também representa a densidade e, por isso, pode mostrar duas regiões de maior concentração, revelando a bimodalidade.

## 9 — Área (`fill_between`)

### Exercício 9.1


```python
meses=np.arange(1,13)
producao=[50,52,55,60,65,70,68,72,78,85,90,95]
plt.figure(figsize=(9,5))
plt.plot(meses,producao,marker='o')
plt.fill_between(meses,producao,0,alpha=.4)
plt.xticks(meses)
plt.xlabel('Mês'); plt.ylabel('Produção')
plt.title('Produção mensal')
plt.show()
```

### Exercício 9.2


```python
x=np.arange(1,6)
media=np.array([100,115,130,150,175])
inferior=np.array([90,100,110,125,140])
superior=np.array([110,130,150,175,210])

plt.figure(figsize=(9,5))
plt.plot(x,media,marker='o',label='Média')
plt.fill_between(x,inferior,superior,alpha=.3,label='Faixa de incerteza')
plt.legend(); plt.title('Projeção de faturamento')
plt.show()
```

**Resposta:** a área preenchida mostra a amplitude de cenários possíveis. Quanto maior a faixa, maior a incerteza, o que ajuda a visualizar risco além da previsão média.

## 10 — Área empilhada (`stackplot`)

### Exercício 10.1


```python
meses=['M1','M2','M3','M4','M5','M6']
solar=[20,25,30,35,45,50]
eolica=[40,45,40,50,55,60]
hidro=[100,95,90,85,80,75]

plt.figure(figsize=(9,5))
plt.stackplot(meses,solar,eolica,hidro,labels=['Solar','Eólica','Hidrelétrica'])
plt.legend(loc='upper left')
plt.ylabel('MW')
plt.title('Geração de energia')
plt.show()
```

### Exercício 10.2 — Resposta


Solar + eólica passam de **60 MW** no primeiro mês para **110 MW** no sexto, enquanto a hidrelétrica cai de **100 MW** para **75 MW**. Assim, a participação relativa das fontes intermitentes cresce ao longo do semestre e termina superior à hidrelétrica.

## 11 — Mapa de calor (`imshow`)

### Exercício 11.1


```python
np.random.seed(42)
matriz=np.random.rand(4,4)
plt.figure(figsize=(6,5))
img=plt.imshow(matriz,cmap='magma')
plt.colorbar(img)
plt.xticks(range(4)); plt.yticks(range(4))
plt.xlabel('Coluna'); plt.ylabel('Linha')
plt.title('Mapa de calor')
plt.show()
```

### Exercício 11.2 — Resposta

2. **Imagens:** imagens digitais são matrizes de pixels. Em RGB, normalmente existem três canais correspondentes a vermelho, verde e azul. `imshow()` exibe essas matrizes visualmente.

## 12 — Contorno (`contour`)

### Exercício 12.1


```python
x=np.linspace(-2,2,200)
y=np.linspace(-2,2,200)
X,Y=np.meshgrid(x,y)
Z=X**2-Y**2

plt.figure(figsize=(7,6))
c=plt.contour(X,Y,Z,levels=15,cmap='coolwarm')
plt.clabel(c,inline=True,fontsize=8)
plt.xlabel('X'); plt.ylabel('Y')
plt.title('Z = X² - Y²')
plt.show()
```

### Exercício 12.2 — Resposta


Linhas de nível muito próximas indicam mudança rápida de Z em uma pequena distância, isto é, maior inclinação/gradiente. Linhas mais espaçadas indicam variação mais gradual.

## 13 — Gráficos 3D

### Exercício 13.1


```python
x=np.linspace(-3,3,100)
y=np.linspace(-3,3,100)
X,Y=np.meshgrid(x,y)
Z=np.sin(X)*np.cos(Y)

fig=plt.figure(figsize=(9,7))
ax=fig.add_subplot(111,projection='3d')
surf=ax.plot_surface(X,Y,Z,cmap='plasma')
ax.set_xlabel('X'); ax.set_ylabel('Y'); ax.set_zlabel('Z')
ax.set_title('Z = sin(X) × cos(Y)')
fig.colorbar(surf,shrink=.6)
plt.show()
```

### Exercício 13.2


```python
np.random.seed(42)
x=np.random.rand(150); y=np.random.rand(150); z=np.random.rand(150)
fig=plt.figure(figsize=(8,6))
ax=fig.add_subplot(111,projection='3d')
ax.scatter(x,y,z)
ax.set_xlabel('X'); ax.set_ylabel('Y'); ax.set_zlabel('Z')
ax.view_init(elev=20,azim=45)
plt.show()
```

**Resposta:** em um eixo criado com `projection='3d'`, usamos `ax.scatter(x, y, z)`, método do eixo tridimensional. `plt.scatter()` é a interface usual para dispersão 2D.

# Desafio Final Integrador


```python
avaliacoes=['P1','P2','P3','P4']
A=[6.5,7.0,6.8,7.2]
B=[7.5,7.8,8.0,8.2]
C=[5.0,6.0,6.5,7.0]
turmas=['Turma A','Turma B','Turma C']
aprovacao=[72,88,64]

fA=[5.0,6.0,6.5,7.0,7.0,7.5,8.0,8.5,9.0]
fB=[7.0,7.5,7.5,8.0,8.0,8.5,8.5,9.0,9.5]
fC=[3.0,4.5,5.5,6.0,6.5,7.0,7.5,8.0,9.5]

fig,ax=plt.subplots(1,3,figsize=(18,5))

ax[0].plot(avaliacoes,A,marker='o',label='A')
ax[0].plot(avaliacoes,B,marker='o',label='B')
ax[0].plot(avaliacoes,C,marker='o',label='C')
ax[0].set_title('Médias por avaliação'); ax[0].legend(); ax[0].grid(True)

ax[1].bar(turmas,aprovacao)
ax[1].set_ylim(0,100); ax[1].set_title('Taxa de aprovação'); ax[1].set_ylabel('%')

ax[2].boxplot([fA,fB,fC],tick_labels=turmas)
ax[2].set_title('Notas finais'); ax[2].set_ylabel('Nota')

plt.tight_layout(); plt.show()

for nome,dados in zip(turmas,[fA,fB,fC]):
    print(nome,
          '| média:',round(np.mean(dados),2),
          '| mediana:',round(np.median(dados),2),
          '| desvio padrão:',round(np.std(dados),2))
```

## Conclusão


A **Turma B** teve o melhor desempenho: maiores médias em P1–P4 e maior aprovação (88%). Também apresenta notas finais mais concentradas, sendo a mais consistente. A **Turma C** apresenta maior variabilidade. Pela regra padrão de 1,5 × IQR, os dados fornecidos podem não produzir pontos marcados como outliers no boxplot; por isso, é mais correto afirmar que C tem maior dispersão do que classificar automaticamente 3,0 ou 9,5 como outliers.
